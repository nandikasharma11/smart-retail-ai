# Module C1 & C2: Unified ML Pipeline Integration

This notebook demonstrates loading all four components (face recognition, product classification, sentiment analysis, chatbot) from a single unified pipeline instance at startup and verifying their end-to-end flow.

```python
import cv2
import numpy as np
import os
import sys
sys.path.append('../')

from app.services.pipeline import pipeline
from app.services import cv_utils

print('Unified pipeline imported successfully.')
```

## 1. Test Face Recognition & Visit Logging

```python
image_path = '../data/lena.jpg'
img = cv_utils.read_image(image_path)

name, sim = pipeline.face_recognizer.recognize_face(img)
print(f"Face recognition query: recognized '{name}' with similarity {sim:.4f}")
if name != 'Unknown' and name != 'No Face Detected':
    timestamp = pipeline.face_recognizer.log_visit(name)
    print(f'Logged visit for {name} at {timestamp}')
```

## 2. Test Product Classification

```python
# We will use the same image just as a shape query to test classification
label, conf = pipeline.product_classifier.predict(img)
print(f'Product classification prediction: {label} (confidence: {conf:.4f})')
```

## 3. Test Sentiment Analysis

```python
text = 'Fantastic service, love it!'
sentiment, conf = pipeline.sentiment_analyzer.predict(text)
print(f"Sentiment analysis of '{text}': {sentiment} (confidence: {conf:.4f})")
```

## 4. Test Chatbot

```python
query = 'when are you open?'
response, tag, conf = pipeline.chatbot.get_response(query)
print(f"Chatbot query: '{query}' -> Response: {response} [Tag: {tag}, conf: {conf:.4f}]")
```
