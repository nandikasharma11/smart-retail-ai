# OpenCV Basics & Face Detection

This notebook demonstrates standard OpenCV operations for image preprocessing and Haar Cascade face detection. It also serves as the prototyping ground for the core CV helper functions.

```python
import cv2
import numpy as np
import os
import urllib.request
import matplotlib.pyplot as plt

%matplotlib inline
```

## 1. Helper function to download sample image

```python
def download_sample_image():
    # Ensure data directory exists
    os.makedirs('../data', exist_ok=True)
    sample_path = '../data/lena.jpg'
    if not os.path.exists(sample_path):
        url = 'https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg'
        print(f'Downloading sample image from {url}...\n')
        urllib.request.urlretrieve(url, sample_path)
        print('Download complete.')
    else:
        print('Sample image already exists.')
    return sample_path

image_path = download_sample_image()
```

## 2. Load and Display Image

```python
image = cv2.imread(image_path)
if image is None:
    print('Error loading image!')
else:
    # OpenCV reads image in BGR format, convert to RGB for displaying with matplotlib
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    plt.imshow(image_rgb)
    plt.title('Original Image (Lena)')
    plt.axis('off')
    plt.show()
```

## 3. Grayscale Conversion

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
plt.imshow(gray_image, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')
plt.show()
```

## 4. Image Resizing
Demonstrating resizing with aspect ratio retention.

```python
h, w = image.shape[:2]
new_width = 256
aspect_ratio = new_width / float(w)
new_height = int(h * aspect_ratio)
resized_image = cv2.resize(image, (new_width, new_height), interpolation=cv2.INTER_AREA)

print(f'Original dimensions: {w}x{h}')
print(f'Resized dimensions: {new_width}x{new_height}')
plt.imshow(cv2.cvtColor(resized_image, cv2.COLOR_BGR2RGB))
plt.title('Resized Image')
plt.axis('off')
plt.show()
```

## 5. Image Blurring

```python
gaussian_blur = cv2.GaussianBlur(image, (15, 15), 0)
median_blur = cv2.medianBlur(image, 15)
bilateral_blur = cv2.bilateralFilter(image, 15, 75, 75)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(cv2.cvtColor(gaussian_blur, cv2.COLOR_BGR2RGB))
axes[0].set_title('Gaussian Blur')
axes[0].axis('off')

axes[1].imshow(cv2.cvtColor(median_blur, cv2.COLOR_BGR2RGB))
axes[1].set_title('Median Blur')
axes[1].axis('off')

axes[2].imshow(cv2.cvtColor(bilateral_blur, cv2.COLOR_BGR2RGB))
axes[2].set_title('Bilateral Blur')
axes[2].axis('off')

plt.show()
```

## 6. Canny Edge Detection

```python
edges = cv2.Canny(gray_image, 50, 150)
plt.imshow(edges, cmap='gray')
plt.title('Canny Edge Detection')
plt.axis('off')
plt.show()
```

## 7. Face Detection with Haar Cascade

```python
cascade_path = os.path.join(cv2.data.haarcascades, 'haarcascade_frontalface_default.xml')
face_cascade = cv2.CascadeClassifier(cascade_path)

faces = face_cascade.detectMultiScale(gray_image, scaleFactor=1.1, minNeighbors=5, minSize=(30, 30))
print(f'Detected {len(faces)} face(s)')

img_faces = image.copy()
for (x, y, w, h) in faces:
    cv2.rectangle(img_faces, (x, y), (x+w, y+h), (0, 255, 0), 2)

plt.imshow(cv2.cvtColor(img_faces, cv2.COLOR_BGR2RGB))
plt.title('Face Bounding Boxes')
plt.axis('off')
plt.show()
```

## 8. Video Capture / Simulation
Since running inside a headless environment or non-interactive notebook makes direct webcam capture `cv2.VideoCapture(0)` fail or block, we write a standard block for video frame processing. If we run it in a headless environment, we simulate a video feed by processing the static image multiple times to represent frames.

```python
def simulate_video_processing(image, num_frames=5):
    print('Simulating video capture frame processing...')
    for i in range(num_frames):
        # Simulate receiving a frame from a video stream
        frame = image.copy()
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        faces = face_cascade.detectMultiScale(gray, 1.1, 5)
        print(f'Frame {i+1}: processed, detected {len(faces)} faces.')
    print('Video capture simulation done.')

simulate_video_processing(image)
```
